#Одновимірний масив

```python
even = np.arange(2, 21, 2)
print('Масив:', even)
print('Сума:', even.sum())
print('Середнє:', even.mean())
print('Мінімум:', even.min())
print('Максимум:', even.max())
print('Помножений на 3:', even * 3)
```

#Двовимірний масив 3×3

```python
m = np.random.randint(1, 11, size=(3, 3))
print('Матриця:\n', m)
print('Сума за стовпцями (axis=0):', m.sum(axis=0))
print('Сума за рядками (axis=1):', m.sum(axis=1))
```
